```python
1.Input the string “Python” as a list of characters from console, delete at least 2 characters, reverse the
resultantstring and print it.
```

```python
# Read input string
s = input().strip()

# Convert string to list of characters
chars = list(s)
del chars[:2]

# Reverse the remaining characters
chars.reverse()

# Convert back to string and print
print("".join(chars))
```

```python
Take two numbers from user and perform at least 4 arithmetic operations on them
```

```python
# Read two numbers
a = int(input())
b = int(input())

# Arithmetic operations
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)

# Handle division safely
if b != 0:
    print("Division:", a / b)
else:
    print("Division: Not possible (division by zero)")
```

```python
3.Use the if statement conditions to write a program to print the letter grade based on an input class score. Use the
grading scheme we are using in this class.
```

```python
# Read class score
score = int(input())

# Determine grade
if score >= 90 and score <= 100:
    print("Grade: A")
elif score >= 80 and score < 90:
    print("Grade: B")
elif score >= 70 and score < 80:
    print("Grade: C")
elif score >= 60 and score < 70:
    print("Grade: D")
elif score >= 0 and score < 60:
    print("Grade: F")
else:
    print("Invalid score")
```
